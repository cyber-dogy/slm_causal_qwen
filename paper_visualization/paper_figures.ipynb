# Paper Figures Notebook

这个 notebook 用于论文图表的一键整理与导出。

建议顺序：
1. 重新整理实验数据
2. 生成所有论文图
3. 预览关键图表与表格


```python
from pathlib import Path
import json
import subprocess
import pandas as pd
from IPython.display import Image, display, Markdown

REPO_ROOT = Path.cwd().resolve().parents[0] if (Path.cwd().name == 'paper_visualization') else Path.cwd().resolve()
if not (REPO_ROOT / 'paper_visualization').exists():
    REPO_ROOT = Path(__file__).resolve().parents[1] if '__file__' in globals() else REPO_ROOT
PV_ROOT = REPO_ROOT / 'paper_visualization'
CACHE_DIR = PV_ROOT / 'cache'
EXPORT_DIR = PV_ROOT / 'exports'
print('PV_ROOT =', PV_ROOT)
```

## 1. 重新整理实验数据

```python
subprocess.run(['python', str(PV_ROOT / 'prepare_paper_data.py')], check=True)
summary = json.loads((CACHE_DIR / 'prepare_summary.json').read_text(encoding='utf-8'))
summary
```

## 2. 生成论文图表

```python
subprocess.run(['python', str(PV_ROOT / 'plot_paper_figures.py')], check=True)
plot_summary = json.loads((EXPORT_DIR / 'plot_summary.json').read_text(encoding='utf-8'))
plot_summary
```

## 3. 查看关键主结果表

```python
data1_main = pd.read_csv(EXPORT_DIR / 'table_data1_main_results.csv')
data2_main = pd.read_csv(EXPORT_DIR / 'table_data2_transfer_results.csv')
display(Markdown('### Data1 Main Results'))
display(data1_main[['paper_label','benchmark','task_mode','split','stage','macro_f1_mean','balanced_accuracy_mean']].sort_values('macro_f1_mean', ascending=False))
display(Markdown('### Data2 Transfer Results'))
display(data2_main[['paper_label','benchmark','task_mode','split','stage','macro_f1_mean','balanced_accuracy_mean']].sort_values('macro_f1_mean', ascending=False))
```

## 4. 预览关键图表

```python
figure_names = [
    'fig01_data1_main_results.png',
    'fig02_qwen_input_ablation.png',
    'fig03_causal_gain.png',
    'fig04_last2_before_after.png',
    'fig05_data2_transfer_best.png',
    'fig06_per_class_recall_heatmap.png',
    'fig07_protocol_distribution.png',
]
for name in figure_names:
    path = EXPORT_DIR / name
    if path.exists():
        display(Markdown(f'### {name}'))
        display(Image(filename=str(path)))
```

## 5. 查看 Qwen 输入消融表

```python
qwen_ablation = pd.read_csv(CACHE_DIR / 'qwen_input_ablation_3class.csv')
qwen_ablation[['experiment_name','input_mode','method','macro_f1_mean','balanced_accuracy_mean']].sort_values('macro_f1_mean', ascending=False)
```

## 6. 后续建议

- 若新增实验，只需重新运行本 notebook 前两步即可刷新图表。
- 若要改论文配色或字体，直接编辑 `plot_paper_figures.py`。
- 若要新增 figure，建议先在 `prepare_paper_data.py` 中补齐 tidy 数据列。
